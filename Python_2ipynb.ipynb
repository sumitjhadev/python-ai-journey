````python
# -----------------------------------
# MINIMUM OF THREE NUMBERS
# -----------------------------------

# This program finds the smallest number
# among three numbers entered by the user.

# -----------------------------------
# TAKING INPUT FROM USER
# -----------------------------------

first_number = int(input("Enter first number: "))
second_number = int(input("Enter second number: "))
third_number = int(input("Enter third number: "))

# -----------------------------------
# CHECKING WHETHER FIRST NUMBER
# IS THE SMALLEST
# -----------------------------------

if first_number < second_number and first_number < third_number:

    print("First number is the smallest:", first_number)

# -----------------------------------
# CHECKING WHETHER SECOND NUMBER
# IS THE SMALLEST
# -----------------------------------

elif second_number < first_number and second_number < third_number:

    print("Second number is the smallest:", second_number)

# -----------------------------------
# IF ABOVE CONDITIONS ARE FALSE,
# THIRD NUMBER IS THE SMALLEST
# -----------------------------------

else:

    print("Third number is the smallest:", third_number)
````

````python
# -----------------------------------
# NUMBER GUESSING GAME
# -----------------------------------

# Importing random module
import random

# Generating a random number between 1 and 100
jackpot = random.randint(1, 100)

# Counter to track number of attempts
counter = 1

# Taking first guess from user
guess = int(input("Enter your guess: "))

# Loop runs until user guesses correctly
while jackpot != guess:

    # Checking whether guess is smaller
    if guess < jackpot:

        print("Guess higher")

    # Checking whether guess is greater
    else:

        print("Guess lower")

    # Taking next guess
    guess = int(input("Enter your guess: "))

    # Increasing attempt counter
    counter = counter + 1

# Runs when user guesses correctly
print("Correct guess")
print("Number of attempts:", counter)
````

````python
# =========================================================
# Program to calculate In-Hand Salary after deductions
# =========================================================

# Step 1: Take salary input from the user
# int() converts the input into integer value
salary = int(input("Enter your annual salary: "))

# =========================================================
# Step 2: Calculate fixed deductions
# =========================================================

# HRA deduction = 10% of salary
hra = salary * 0.10

# DA deduction = 5% of salary
da = salary * 0.05

# PF deduction = 3% of salary
pf = salary * 0.03

# =========================================================
# Step 3: Calculate tax according to salary slabs
# =========================================================

# If salary is below 5 lakh
if salary < 500000:
    tax = 0

# If salary is between 5 lakh and 10 lakh
elif 500000 <= salary <= 1000000:
    tax = salary * 0.10

# If salary is between 10 lakh and 20 lakh
elif 1000000 < salary <= 2000000:
    tax = salary * 0.20

# If salary is above 20 lakh
else:
    tax = salary * 0.30

# =========================================================
# Step 4: Calculate final in-hand salary
# =========================================================

# Final salary after all deductions
in_hand_salary = salary - hra - da - pf - tax

# =========================================================
# Step 5: Print all deductions and final salary
# =========================================================

print("\n========== Salary Breakdown ==========")

print("Original Salary :", salary)

print("HRA Deduction   :", hra)

print("DA Deduction    :", da)

print("PF Deduction    :", pf)

print("Tax Deduction   :", tax)

print("--------------------------------------")

print("In-Hand Salary  :", in_hand_salary)

print("======================================")
````

### `Problem 2`: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.


````python
# =====================================================
# Program to check whether three angles form a triangle
# =====================================================

# Step 1: Take three angle inputs from the user

angle1 = int(input("Enter first angle: "))
angle2 = int(input("Enter second angle: "))
angle3 = int(input("Enter third angle: "))

# =====================================================
# Step 2: Check triangle condition
# =====================================================

# A triangle is possible only if:
# 1. Sum of all angles is 180
# 2. All angles are greater than 0

if angle1 > 0 and angle2 > 0 and angle3 > 0 and (angle1 + angle2 + angle3 == 180):

    print("Triangle can be formed")

else:

    print("Triangle cannot be formed")
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# =====================================================
# Program to check Profit or Loss
# =====================================================

# Step 1: Take input from the user

cp = int(input("Enter the Cost Price: "))
sp = int(input("Enter the Selling Price: "))

# =====================================================
# Step 2: Check for profit or loss
# =====================================================

# If selling price is greater than cost price
if sp > cp:

    profit = sp - cp

    print("You made a Profit of:", profit)

# If cost price is greater than selling price
elif cp > sp:

    loss = cp - sp

    print("You made a Loss of:", loss)

# If both are equal
else:

    print("No Profit No Loss")
````

### `Problem 4`: Write a menu-driven program -
1. cm to ft
2. km to miles
3. USD to INR
4. exi


````python
# menu driven program

a = int(input("""Hello Sir, How can I help you today

1. cm to feet
2. km to miles
3. USD to INR
4. exit

Enter your choice:
"""))

# option 1

if a == 1:

    cm = float(input("Enter centimeter: "))

    feet = cm / 30.48

    print("Feet =", feet)

# option 2

elif a == 2:

    km = float(input("Enter kilometer: "))

    miles = km / 1.609

    print("Miles =", miles)

# option 3

elif a == 3:

    usd = float(input("Enter USD: "))

    inr = usd * 98

    print("INR =", inr)

# option 4

elif a == 4:

    print("Program exited")

# invalid choice

else:

    print("Invalid option")
````

### `Problem 5` - Exercise 12: Display Fibonacci series up to 10 terms.


````python
# =====================================================
# Program to print Fibonacci Series
# =====================================================

# taking number of terms from user

n = int(input("Enter the number of terms: "))

# first two fibonacci numbers

a = 0
b = 1

# heading

print("\nFibonacci Series:\n")

# loop will run n times

for i in range(n):

    # print current fibonacci number

    print(a)

    # find next fibonacci number

    c = a + b

    # shifting values forward

    a = b

    b = c
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# =====================================================
# Program to find factorial of a number
# =====================================================

# taking input from user

n = int(input("Enter a number: "))

# factorial starts from 1

fact = 1

# loop from 1 to n

for i in range(1, n + 1):

    # multiplying numbers one by one

    fact = fact * i

# printing final factorial

print("Factorial =", fact)
````

```

`Output:`

```bash

````python
num = 123

reversed_num = 0

while num!= 0 :
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num = num // 10

print("reverse of the given numbers is :", reversed_num)
````
